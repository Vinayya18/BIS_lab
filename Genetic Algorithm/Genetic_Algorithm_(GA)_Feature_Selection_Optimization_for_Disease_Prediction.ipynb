```python

import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

data = load_breast_cancer()
X = data.data
y = data.target

population_size = 20
generations = 30
mutation_rate = 0.1
crossover_rate = 0.8

n_features = X.shape[1]

def fitness(chromosome):
    selected = np.where(chromosome == 1)[0]

    if len(selected) == 0:
        return 0

    X_train, X_test, y_train, y_test = train_test_split(
        X[:, selected], y, test_size=0.2, random_state=42
    )

    model = KNeighborsClassifier(n_neighbors=5)
    model.fit(X_train, y_train)

    prediction = model.predict(X_test)

    accuracy = accuracy_score(y_test, prediction)

    return accuracy - 0.01 * len(selected) / n_features

population = np.random.randint(
    0, 2, (population_size, n_features)
)

best_solution = None
best_fitness = -1

for generation in range(generations):

    scores = np.array([fitness(c) for c in population])

    best_index = np.argmax(scores)

    if scores[best_index] > best_fitness:
        best_fitness = scores[best_index]
        best_solution = population[best_index].copy()

    new_population = []

    for i in range(population_size):

        parents = np.random.choice(
            population_size, 2, replace=False
        )

        parent1 = population[parents[0]]
        parent2 = population[parents[1]]

        point = np.random.randint(1, n_features)

        if np.random.random() < crossover_rate:
            child = np.concatenate((
                parent1[:point], parent2[point:]
            ))
        else:
            child = parent1.copy()

        mutation = np.random.random(n_features) < mutation_rate
        child[mutation] = 1 - child[mutation]

        new_population.append(child)

    population = np.array(new_population)

    print("Generation:", generation + 1,
          "Best Fitness:", round(best_fitness, 4))

print("Best Accuracy Fitness:", best_fitness)
print("Selected Features:", np.where(best_solution == 1)[0])
print("Number of Selected Features:", np.sum(best_solution))
```
